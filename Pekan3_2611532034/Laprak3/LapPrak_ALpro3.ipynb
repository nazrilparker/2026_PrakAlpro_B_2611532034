## 2.1 Dasar Teori

## 2.1.1 Pengertian Operator

Operator adalah simbol atau tanda khusus yang digunakan dalam pemrograman untuk melakukan suatu operasi terhadap satu atau lebih nilai (operand). Operator dapat digunakan untuk melakukan berbagai proses, seperti perhitungan matematika, memberikan nilai pada variabel, membandingkan nilai, menggabungkan kondisi, memeriksa identitas atau keanggotaan suatu data, dan mengolah data dalam bentuk biner.

Dalam praktikum ini, operator yang dipelajari meliputi operator aritmatika, assignment, perbandingan, logika, membership, identity, dan bitwise.

### 2.1.2 Operator Aritmatika

Operator aritmatika adalah operator yang digunakan untuk melakukan operasi atau perhitungan matematika terhadap nilai atau variabel. Berikut operator aritmatika dalam Python:

●  + untuk penjumlahan

●  - untuk pengurangan

●  * untuk perkalian

●  / untuk pembagian

●  // untuk pembagian bilangan bulat

●  % untuk mencari sisa pembagian

●  ** untuk perpangkatan

Dalam program transaksi, operator aritmatika dapat digunakan untuk menghitung besar diskon, total pembayaran, rata-rata harga barang, dan sisa pembagian.

### 2.1.3 Operator Assignment

Operator assignment digunakan untuk memasukkan atau mengubah nilai pada sebuah variabel. Operator dasarnya adalah =. Python juga menyediakan augmented assignment yang dapat digunakan untuk mengubah nilai variabel sekaligus melakukan operasi tertentu.

Contohnya:

nilai_2034 = 25
nilai_2034 += 10

Setelah dijalankan, nilai nilai_2034 menjadi 35.

Beberapa operator assignment yang sering digunakan adalah +=, -=, *=, dan /=.

### 2.1.4 Operator Perbandingan

Operator perbandingan digunakan untuk melihat hubungan antara dua nilai. Hasil yang diberikan berupa True atau False, tergantung pada kondisi yang dibandingkan.

Beberapa operator perbandingan yang digunakan dalam Python yaitu:

●  == untuk mengecek apakah nilainya sama

●  != untuk mengecek apakah nilainya berbeda

●  > untuk mengecek nilai lebih besar

●  < untuk mengecek nilai lebih kecil

●  >= untuk mengecek nilai lebih besar atau sama dengan

●  <= untuk mengecek nilai lebih kecil atau sama dengan

Pada program transaksi, operator ini digunakan untuk mengecek apakah total belanja sudah mencapai Rp200.000 dan apakah jumlah barang sudah mencapai minimal 3 barang.

### 2.1.5 Operator Logika

Operator logika digunakan untuk menggabungkan beberapa kondisi dalam sebuah program. Python memiliki tiga operator logika yang sering digunakan, yaitu and, or, dan not.

Operator and akan menghasilkan True jika semua kondisi bernilai True. Operator or menghasilkan True jika salah satu kondisi bernilai True. Sedangkan not digunakan untuk membalik nilai Boolean.

Contohnya:

member_2034 = True
belanja_2034 = True

hasil_2034 = member_3002 and belanja_2034
Hasil dari operasi tersebut adalah True karena kedua kondisi memiliki nilai True.

### 2.1.6 Operator Keanggotaan

Operator membership digunakan untuk memeriksa apakah suatu nilai terdapat di dalam suatu koleksi data seperti list, tuple, atau string.

Operator membership yang digunakan dalam Python adalah:

in not in

Contohnya:

kode_2034 = "HEMAT10"  
daftar_2034 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]

hasil_2034 = kode_2034 in daftar_2034

Hasilnya adalah True karena kode HEMAT10 terdapat di dalam daftar.

### 2.1.7 Operator Identitas (Identity)

Operator identity pada Python berfungsi untuk mengetahui apakah dua variabel menunjuk ke objek yang sama di dalam memori. Jadi, yang diperiksa bukan hanya isi datanya, tetapi apakah keduanya benar-benar menggunakan objek yang sama.

Ada dua operator identity, yaitu:

●  is digunakan untuk mengecek apakah dua variabel merupakan objek yang sama.

●  is not digunakan untuk mengecek apakah dua variabel merupakan objek yang berbeda.

### 2.1.8 Operator Bitwise

Operator bitwise digunakan untuk melakukan operasi terhadap bilangan dalam representasi biner. Operator ini dapat digunakan untuk menggabungkan maupun memeriksa status tertentu yang direpresentasikan menggunakan bit.

Beberapa operator bitwise yang digunakan dalam praktikum adalah:

Operator Fungsi & AND bitwise | OR bitwise ^ XOR bitwise << Left shift

Right shift

Pada studi kasus transaksi, setiap kondisi pelanggan direpresentasikan menggunakan nilai bit:

Nilai Bit Kondisi 0001 Pelanggan merupakan member 0010 Total belanja ≥ Rp200.000 0100 Jumlah barang ≥ 3 1000 Kode promo tersedia

Operator OR digunakan untuk menggabungkan beberapa kondisi menjadi satu kode status. Operator AND digunakan untuk memeriksa kondisi tertentu dari kode status. Sementara itu, XOR dapat digunakan untuk membandingkan perbedaan antara kode transaksi dan kode referensi.

## 2.2 Kode Program

### 2.2.1 Program "aritmatika_2611532034"

```python
# Buat file dengan nama aritmatika_2611532034.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input ()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_2034 = int(input("Masukkan angka-1: "))
angka2_2034 = int(input("Masukkan angka-2: "))

# Penjumlahan
hasil = angka1_2034 + angka2_2034
print("\nOperator Penjumlahan")
print("Hasil =", hasil)

# Pengurangan
hasil = angka1_2034 - angka2_2034
print("\nOperator Pengurangan")
print("Hasil =", hasil)

# Perkalian
hasil = angka1_2034 * angka2_2034
print("\nOperator Perkalian")
print("Hasil =", hasil)

# Pembagian, Pembagian bulat, dan sisa bagi
if angka2_2034 != 0:
    hasil = angka1_2034 / angka2_2034
    print("\nOperator Pembagian")
    print("Hasil =", hasil)

    hasil = angka1_2034 // angka2_2034
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil)

    hasil = angka1_2034 % angka2_2034
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil)
else:
    print("Angka kedua tidak boleh bernilai 0")

# Pangkat
hasil = angka1_2034 ** angka2_2034
print("\nOperator Pangkat")
print("Hasil =", hasil)
```

**Penjelasan :**
1. Pada bagian awal, program meminta pengguna memasukkan dua bilangan. Kedua nilai tersebut disimpan dalam variabel `angka1_2034` dan `angka2_2034` menggunakan `int(input())` agar dibaca sebagai bilangan bulat.

2. Operator `+` digunakan untuk menjumlahkan kedua bilangan, kemudian hasilnya disimpan dalam variabel `hasil` dan ditampilkan menggunakan `print()`.

3. Operator `-` digunakan untuk menghitung selisih antara `angka1_2034` dan `angka2_2034`.

4. Operator `*` digunakan untuk mengalikan kedua bilangan.

5. Sebelum melakukan pembagian, program memeriksa apakah `angka2_2034` tidak sama dengan 0 menggunakan operator `!=`.

6. Jika nilainya bukan 0, operator `/` digunakan untuk melakukan pembagian.

7. Operator `//` digunakan untuk memperoleh hasil pembagian bulat tanpa angka di belakang koma.

8. Operator `%` digunakan untuk mencari sisa pembagian.

9. Jika `angka2_2034` bernilai 0, program tidak melakukan operasi pembagian dan menampilkan pesan bahwa angka kedua tidak boleh bernilai 0.

10. Terakhir, operator `**` digunakan untuk menghitung perpangkatan `angka1_2034` dengan `angka2_2034`.

### 2.2.2 Program "assignment_2611532034"

```python
# Buat file dengan nama assignment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input ()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam Python

angka1_2034 = int(input("Masukkan angka-1: "))
angka2_2034 = int(input("Masukkan angka-2: "))

print("\nNilai awal angka1_2034 =", angka1_2034)
print("Nilai awal angka2_2034 =", angka2_2034)

# Assignment biasa
hasil = angka1_2034
print("\nAssignment biasa (=)")
print("Hasil =", hasil)

# Assignment penambahan
hasil = angka1_2034
hasil += angka2_2034
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil)

# Assignment pengurangan
hasil = angka1_2034
hasil -= angka2_2034
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil)

# Assignment perkalian
hasil = angka1_2034
hasil *= angka2_2034
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_2034 != 0:
    hasil = angka1_2034
    hasil /= angka2_2034
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil)

    #Operator tambahan
    hasil = angka1_2034
    hasil //= angka1_2034
    print("\nAssignment pembagian bulat (//=)")
    print("hasil =", angka1_2034)
    hasil = angka1_2034
    hasil %= angka2_2034
    print("\nAssignment sisa bagi (%=)")
    print ("hasil =", hasil)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0")

# Operator tambahan: assignment perpangkatan
hasil = angka1_2034
hasil **= angka2_2034
print("\nAssignment perpangkatan (**=)")
print("hasil =", hasil)
```

**Penjelasan :**

1. Bagian awal program menerima dua bilangan dari pengguna melalui `input()`, kemudian mengonversinya menjadi tipe data integer dan menampilkan nilai awal `angka1_2034` serta `angka2_2034`.

3. Operator `+=` digunakan untuk menambahkan nilai `angka2_2034` ke dalam `hasil`.

4. Operator `-=` digunakan untuk mengurangi nilai `hasil` dengan `angka2_2034`.

5. Operator `*=` digunakan untuk mengalikan nilai `hasil` dengan `angka2_2034`.

6. Operator `!=` digunakan untuk memeriksa apakah `angka2_2034` tidak bernilai nol sebelum operasi pembagian dilakukan.

7. Operator `/=` digunakan untuk membagi nilai `hasil` dengan `angka2_2034`.

8. Operator `//=` digunakan untuk melakukan pembagian bulat. Pada program ini, `hasil` dibagi dengan `angka1_2034`.

9. Operator `%=` digunakan untuk mencari sisa pembagian `hasil` dengan `angka2_2034`.

10. Jika `angka2_2034` bernilai nol, program menampilkan pesan bahwa pembagian tidak dapat dilakukan.

11. Operator `**=` digunakan untuk menghitung perpangkatan dengan memangkatkan `hasil` menggunakan `angka2_2034`.

### 2.2.3 Program "bitwise_2611532034"

```python
# Buatlah file dengan nama bitwise_2611533002.py
# Nama variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator bitwise dalam Python

print("\n==================================")
print( "OPERATOR BITWISE" )
print("==================================")

angka1_2034 = int(input("Masukkan angka bitwise-1: "))
angka2_2034 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner:")
print("Angka 1:", angka1_2034, "-| biner", bin(angka1_2034))
print("Angka 2:", angka2_2034, "-| biner", bin(angka2_2034))

# Bitwise AND
hasil_2034 = angka1_2034 & angka2_2034
print("\nBitwise AND (&)")
print(angka1_2034, "&", angka2_2034, "=", hasil_2034)
print("Biner hasil =", bin(hasil_2034))
print("Biner hasil (8 bit) =", format(hasil_2034, '08b'))

# Bitwise OR
hasil_2034 = angka1_2034 | angka2_2034
print("\nBitwise OR (|)")
print(angka1_2034, "|", angka2_2034, "=", hasil_2034)
print("Biner hasil =", bin(hasil_2034))
print("Biner hasil (8 bit) =", format(hasil_2034, '08b'))

# Bitwise XOR
hasil_2034 = angka1_2034 ^ angka2_2034
print("\nBitwise XOR (^)")
print(angka1_2034, "^", angka2_2034, "=", hasil_2034)
print("Biner hasil =", bin(hasil_2034))
print("Biner hasil (8 bit) =", format(hasil_2034, '08b'))

# Bitwise NOT
hasil_2034 = ~angka1_2034
print("\nBitwise NOT (~)")
print("~", angka1_2034, "=", hasil_2034)
print("Biner hasil =", bin(hasil_2034))
print("Biner hasil (8 bit) =", format(hasil_2034, '08b'))

# Bitwise geser kiri
jumlah_geser_2034 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2034 = angka1_2034 << jumlah_geser_2034
print("\nBitwise Geser Kiri (<<)")
print(angka1_2034, "<<", jumlah_geser_2034, "=", hasil_2034)
print("Biner hasil =", bin(hasil_2034))
print("Biner hasil (8 bit) =", format(hasil_2034, '08b'))

# Bitwise geser kanan
hasil_2034 = angka1_2034 >> jumlah_geser_2034
print("\nBitwise Geser Kanan (>>)")
print(angka1_2034, ">>", jumlah_geser_2034, "=", hasil_2034)
print("Biner hasil =", bin(hasil_2034))
print("Biner hasil (8 bit) =", format(hasil_2034, '08b'))
```

### 2.2.4 Program "lainnya_2611532034"

```python
# Buatlah file dengan nama lainnya_2611532034.py
# Nama variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas dalam Python

print("==================================")
print("1. OPERATOR KEANGGOTAAN")
print("==================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_2034 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
input_data_2034 = [int(angka.strip()) for angka in input_data_2034.split(",")]

nilai_dicari_2034 = int(input("Masukkan angka yang ingin dicari: "))

# Operator In
hasil_2034 = nilai_dicari_2034 in input_data_2034
print("\nOperator keanggotaan IN")
print(nilai_dicari_2034, "in", input_data_2034, "=", hasil_2034)

# Operator Not In
hasil_2034 = nilai_dicari_2034 not in input_data_2034
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_2034, "not in", input_data_2034, "=", hasil_2034)

print("\n==================================")
print("2. OPERATOR IDENTITAS")
print("==================================")

# objek1 menggunakan list dari input pengguna
objek1_2034 = input_data_2034

# objek2 menggunakan list dari input pengguna
objek2_2034 = objek1_2034

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_2034 = input_data_2034.copy()

print("objek1_2034 =", objek1_2034)
print("objek2_2034 =", objek2_2034)
print("objek3_2034 =", objek3_2034)

# Operator Is
hasil_2034 = objek1_2034 is objek2_2034
print("\nOperator identitas IS")
print("objek1_2034 is objek2_2034 =", hasil_2034)

# Operator Is Not
hasil_2034 = objek1_2034 is not objek3_2034
print("\nOperator identitas IS NOT")
print("objek1_2034 is not objek3_2034 =", hasil_2034)

# Membandingkan identitas dan nilai
print("\nMembandingkan identitas dan nilai:")
print("objek1_2034 is objek3_2034 =", objek1_2034 is objek3_2034)
print("objek1_2034 == objek3_2034 =", objek1_2034 == objek3_2034)
```

### 2.2.5 Program "logika_2611532034"

```python
# Buat file dengan nama logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a1_1234
# Program ini menggunakan fungsi input ()
# Program operator logika dalam Python

# Memasukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_2034 = input("Input nilai boolean-1 (True/False): ").strip().lower() == "true"
a2_2034 = input("Input nilai boolean-2 (True/False): ").strip().lower() == "true"

print("\nA1 = ", a1_2034)
print("A2 = ", a2_2034)

# Konjungsi: bernilai True jika keduanya True
hasil = a1_2034 and a2_2034
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil)

# Disjungsi: bernilai True jika salah satunya True
hasil = a1_2034 or a2_2034
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil)

# Negasi A1: membalik nilai A1
hasil = not a1_2034
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil)

# Negasi A2: membalik nilai A2
hasil = not a2_2034
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil)

# XOR: bernilai True jika kedua nilai berbeda
hasil = a1_2034 != a2_2034
print ("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil)
```

**Penjelasan :**

1. Baris 8–9 digunakan untuk menerima nilai boolean dari pengguna. Input diubah menjadi huruf kecil menggunakan `lower()`, kemudian dibandingkan dengan `"true"`.

2. Baris 11–12 menampilkan nilai `a1_2034` dan `a2_2034`.

3. Baris 15–17 menggunakan operator `and`. Hasilnya `True` jika kedua nilai bernilai `True`.

4. Baris 19–21 menggunakan operator `or`. Hasilnya `True` jika salah satu atau kedua nilai bernilai `True`.

5. Baris 23–25 menggunakan operator `not` pada `a1_2034` untuk membalik nilai boolean.

6. Baris 27–29 menggunakan operator `not` pada `a2_2034`.

7. Baris 31–33 menggunakan operator `!=` sebagai XOR. Hasilnya `True` jika nilai `a1_2034` dan `a2_2034` berbeda.

8. Secara keseluruhan, program ini menunjukkan penggunaan operator logika `and`, `or`, `not`, dan XOR untuk mengolah dua nilai boolean.

### 2.2.6 Program "perbandingan_2611532034"

```python
# Buat file dengan nama perbandingan_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input ()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam Python

angka1_2034 = int(input("Masukkan angka-1: "))
angka2_2034 = int(input("Masukkan angka-2: "))

# Lebih besar dari
hasil = angka1_2034 > angka2_2034
print("\nOperator lebih besar dari")
print("angka1_2034 > angka2_2034 =", hasil)

# Lebih kecil dari
hasil = angka1_2034 < angka2_2034
print("\nOperator lebih kecil dari")
print("angka1_2034 < angka2_2034 =", hasil)

# Lebih besar dari atau sama dengan
hasil = angka1_2034 >= angka2_2034
print("\nOperator lebih besar dari atau sama dengan")
print("angka1_2034 >= angka2_2034 =", hasil)

# Lebih kecil dari atau sama dengan
hasil = angka1_2034 <= angka2_2034
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1_2034 <= angka2_2034 =", hasil)

# Sama dengan
hasil = angka1_2034 == angka2_2034
print("\nOperator sama dengan")
print("angka1_2034 == angka2_2034 =", hasil)

# Tidak sama dengan
hasil = angka1_2034 != angka2_2034
print("\nOperator tidak sama dengan")
print("angka1_2034 != angka2_2034 =", hasil)

# Tambahan: perbandingan berantai dalam Python
hasil = 0 < angka1_2034 < 100
print("\nPerbandingan berantai")
print("0 < angka1_2034 < 100 =", hasil)

hasil = 0 < angka2_2034 < 100
print("0 < angka2_2034 < 100 =", hasil)
```

**Penjelasan :**

1. Baris 8–9 digunakan untuk meminta dua angka dari pengguna. Fungsi `input()` menerima data, kemudian `int()` mengubahnya menjadi bilangan bulat dan menyimpannya pada `angka1_2034` dan `angka2_2034`.

5. Baris 24–26 mengecek apakah `angka1_2034` lebih kecil atau sama dengan `angka2_2034` menggunakan operator `<=`.

6. Baris 28–30 membandingkan apakah kedua angka memiliki nilai yang sama menggunakan operator `==`.

7. Baris 32–34 mengecek apakah kedua angka memiliki nilai yang berbeda menggunakan operator `!=`.

8. Baris 36–38 menggunakan perbandingan berantai `0 < angka1_2034 < 100` untuk mengecek apakah nilai `angka1_2034` berada antara 0 dan 100.

## Tugas Praktikum

**Soal : Sistem Simulasi Transaksi dan Validasi Akses Toko. Anda diminta untuk membuat sebuah program kasir sederhana yang digunakan untuk menghitung total transaksi pelanggan sekaligus melakukan pemeriksaan terhadap status pelanggan dan hak akses terhadap promo tertentu. Program menerima data transaksi dan informasi pelanggan, kemudian menggunakan berbagai jenis operator Python untuk menentukan total pembayaran, status diskon, kelayakan promo, serta hak akses pelanggan. Seluruh proses harus dibuat dalam satu alur program yang utuh.**



**Kode Jawaban :**

a. Kode Program "Tugas3_2611532034"

```python
print("=== SISTEM TRANSAKSI TOKO ===")
print(" ")
print(" ")

# INPUT DATA PELANGGAN
nama_2034 = input("Masukkan Nama Pelanggan : ")
print(" ")
status_2034 = input("Masukkan Status Pelanggan (member/nonmember) : ")
print(" ")
total_belanja_2034 = float(input("Masukkan Total Belanja : Rp"))
print(" ")
jumlah_barang_2034 = int(input("Masukkan Jumlah Barang : "))
print(" ")
kode_promo_2034 = input("Masukkan Kode Promo : ")

# OPERATOR PERBANDINGAN
syarat_belanja_2034 = total_belanja_2034 >= 200000
syarat_barang_2034 = jumlah_barang_2034 >= 3
status_member_2034 = status_2034 == "member"

# DAFTAR KODE PROMO
daftar_promo_2034 = ["HEMAT10", "MURAH10", "HAPPYSHOP"]

# OPERATOR KEANGGOTAAN
promo_tersedia_2034 = kode_promo_2034 in daftar_promo_2034
promo_tidak_tersedia_2034 = kode_promo_2034 not in daftar_promo_2034

# OPERATOR LOGIKA 
diskon_member_2034 = status_member_2034 and syarat_belanja_2034

promo_didapatkan_2034 = promo_tersedia_2034 and (syarat_belanja_2034 or syarat_barang_2034)

bukan_member_2034 = not status_member_2034

# OPERATOR ARITMATIKA
if diskon_member_2034:
    persentase_diskon_2034 = 0.10
else:
    persentase_diskon_2034 = 0.05 if syarat_belanja_2034 else 0

besar_diskon_2034 = total_belanja_2034 * persentase_diskon_2034

total_pembayaran_2034 = total_belanja_2034 - besar_diskon_2034

if jumlah_barang_2034 > 0:
    rata_rata_barang_2034 = total_belanja_2034 / jumlah_barang_2034
else:
    rata_rata_barang_2034 = 0

sisa_pembagian_2034 = int(total_belanja_2034) % jumlah_barang_2034 if jumlah_barang_2034 > 0 else 0

# OPERATOR PENUGASAN
poin_2034 = 0
if status_member_2034:
    poin_2034 += int(total_pembayaran_2034 // 10000)

jumlah_barang_tersisa_2034 = jumlah_barang_2034
if promo_didapatkan_2034:
    jumlah_barang_tersisa_2034 -= 1

# OPERATOR IDENTITY (IDENTITAS)
kode_1_2034 = ["HEMAT10"]
kode_2_2034 = ["MURAH10"]

nilai_sama_2034 = kode_1_2034 == kode_2_2034
objek_sama_2034 = kode_1_2034 is kode_2_2034
objek_berbeda_2034 = kode_1_2034 is not kode_2_2034

# OPERATOR BITWISE
kode_member_2034 = 1 if status_member_2034 else 0
kode_belanja_2034 = 2 if syarat_belanja_2034 else 0
kode_barang_2034 = 4 if syarat_barang_2034 else 0
kode_promo_2034 = 8 if promo_tersedia_2034 else 0

kode_status_2034 = (
    kode_member_2034
    | kode_belanja_2034
    | kode_barang_2034
    | kode_promo_2034
)

cek_member_2034 = kode_status_2034 & 1
cek_belanja_2034 = kode_status_2034 & 2
cek_barang_2034 = kode_status_2034 & 4
cek_promo_2034 = kode_status_2034 & 8

kode_referensi_2034 = 11
perbandingan_status_2034 = kode_status_2034 ^ kode_referensi_2034

kode_shift_2034 = kode_status_2034 << 1


member_access_2034 = bool(cek_member_2034)
promo_access_2034 = bool(cek_promo_2034)
free_shipping_access_2034 = syarat_belanja_2034 and syarat_barang_2034

# OUTPUT
print("\n=== DATA TRANSAKSI ===")
print("Nama Pelanggan       : ", nama_2034)
print(" ")
print("Status Pelanggan     : ", status_2034)
print(" ")
print("Total Belanja        : Rp", total_belanja_2034)
print(" ")
print("Jumlah Barang        : ", jumlah_barang_2034)
print(" ")
print("Kode Promo           : ", kode_promo_2034)
print(" ")

print("\n=== HASIL VALIDASI ===")
print(" ")
print("Belanja >= Rp200000  :", syarat_belanja_2034)
print(" ")
print("Jumlah Barang >= 3   :", syarat_barang_2034)
print(" ")
print("Status Member        :", status_member_2034)
print(" ")
print("Kode Promo Tersedia  :", promo_tersedia_2034)
print(" ")
print("Mendapatkan Diskon   :", diskon_member_2034)
print(" ")
print("Mendapatkan Promo    :", promo_didapatkan_2034)
print(" ")

print("\n=== HASIL PERHITUNGAN ===")
print(" ")
print("Besarnya Diskon      : Rp", besar_diskon_2034)
print(" ")
print("Total Pembayaran     : Rp", total_pembayaran_2034)
print(" ")
print("Rata-rata Harga      : Rp", rata_rata_barang_2034)
print(" ")
print("Sisa Pembagian       :", sisa_pembagian_2034)
print(" ")

print("\n=== HAK AKSES PELANGGAN ===")
print("Kode Hak Akses       :", format(kode_status_2034, "04b"))
print("Member Access        :", member_access_2034)
print("Promo Access         :", promo_access_2034)
print("Free Shipping Access :", free_shipping_access_2034)
print("Poin Pelanggan       :", poin_2034)

print("\n=== OPERATOR IDENTITAS ===")
print("kode_1 == kode_2     :", nilai_sama_2034)
print("kode_1 is kode_2     :", objek_sama_2034)
print("kode_1 is not kode_2 :", objek_berbeda_2034)

print("\n=== OPERASI BITWISE ===")
print("Kode Status Transaksi")
print("0001 | 0010 | 0100 | 1000")

print("Kode Biner           :", format(kode_status_2034, "04b"))
print("Kode Desimal         :", kode_status_2034)

print("\n=== PEMERIKSAAN STATUS ===")

print("\nCek Member")
print(format(kode_status_2034, "04b"), "& 0001")
print("Hasil Biner         :", format(cek_member_2034, "04b"))
print("Hasil Desimal       :", cek_member_2034)

print("\nCek Belanja")
print(format(kode_status_2034, "04b"), "& 0010")
print("Hasil Biner         :", format(cek_belanja_2034, "04b"))
print("Hasil Desimal       :", cek_belanja_2034)

print("\nCek Jumlah Barang")
print(format(kode_status_2034, "04b"), "& 0100")
print("Hasil Biner         :", format(cek_barang_2034, "04b"))
print("Hasil Desimal       :", cek_barang_2034)

print("\nCek Promo")
print(format(kode_status_2034, "04b"), "& 1000")
print("Hasil Biner         :", format(cek_promo_2034, "04b"))
print("Hasil Desimal       :", cek_promo_2034)

print("\n=== PERBANDINGAN STATUS (XOR) ===")
print("Kode Transaksi   :", format(kode_status_2034, "04b"))
print("Kode Referensi      :", format(kode_referensi_2034, "04b"))
print(format(kode_status_2034, "04b"), "^",
    format(kode_referensi_2034, "04b"))
print("Hasil Biner         :", format(perbandingan_status_2034, "04b"))
print("Hasil Desimal       :", perbandingan_status_2034)

print("\n=== SHIFT ===")
print(format(kode_status_2034, "04b"), "<< 1")
print("Hasil Biner         :", format(kode_shift_2034, "b"))
print("Hasil Desimal       :", kode_shift_2034)
print(" ")
print(" ")

print("=== SELESAI ===")
```

Program menggunakan operator aritmatika seperti perkalian (*) untuk menghitung diskon, pengurangan (-) untuk mendapatkan total pembayaran setelah diskon, pembagian (/) untuk menghitung rata-rata harga barang, dan modulus (%) untuk mencari sisa pembagian jumlah barang.
