## 2.1 Dasar Teori

    1. Pengenalan Bahasa Python
Python merupakan salah satu bahasa pemrograman tingkat tinggi (high-level programming language) yang bersifat general-purpose, artinya dapat digunakan untuk berbagai keperluan pengembangan perangkat lunak, mulai dari aplikasi desktop, web, analisis data, kecerdasan buatan (artificial intelligence), hingga machine learning. Python memiliki bentuk penulisan yang relatif sederhana sehingga mudah dibaca dan dipelajari. Python diciptakan oleh Guido van Rossum dan pertama kali dirilis pada tahun 1991.

    2. Variabel
Variabel adalah suatu identitas atau nama yang digunakan untuk menyimpan suatu nilai (data) di dalam memori komputer, di mana nilai tersebut dapat diakses, diubah, maupun digunakan kembali selama program dijalankan. Variabel dapat dianalogikan sebagai sebuah "wadah" yang menampung data tertentu. Dalam penulisannya, variabel memiliki beberapa aturan seperti tidak boleh mengandung spasi atau simbol khusus seperti @, #, % dan hanya boleh terdiri dari huruf, angka, dan garis bawah.

    3. Tipe Data
Tipe data menunjukkan bentuk atau jenis nilai yang terdapat pada suatu variabel. Python memiliki beberapa tipe data bawaan yang sering digunakan antara lain:
- Integer (int): tipe data untuk bilangan bulat, baik positif maupun negatif, tanpa mengandung angka desimal.

- Float (float): tipe data untuk bilangan pecahan atau desimal.

- String (str): tipe data untuk kumpulan karakter atau teks, yang dituliskan di antara tanda kutip tunggal ('...') atau tanda kutip ganda ("...").

- Boolean (bool): menyimpan nilai logika yang terdiri dari True dan False.


## 2.2 Kode Program

### 2.2.1 Program "Boolean_2611532034"

```python
# Buat file dengan nama Boolean_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasi variabel dengan tipe data Boolean
is_lulus_2034 = True
is_cumlaude_2034 = True

# Menggunakan Boolean
nilai_2034 = 85
batas_lulus_2034 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan_2034 = nilai_2034 >= batas_lulus_2034 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai_2034)
print("Apakah Lulus?:", status_kelulusan_2034)
if is_lulus_2034 and is_cumlaude_2034:
    print("Selamat Anda lulus dengan predikat cumlaude!")
```

**Penjelasan :**
1. Dua variabel pertama berisi nilai Boolean untuk menunjukkan status lulus dan status cumlaude.
2. nilai_2034 menyimpan nilai mahasiswa, sedangkan batas_lulus_2034 menyimpan nilai minimum untuk dinyatakan lulus.
3. Perbandingan kedua nilai tersebut menghasilkan True atau False pada variabel status kelulusan.
4. Hasil akhirnya ditampilkan ke layar menggunakan beberapa perintah print().

### 2.2.2 Program "Input_2611532034"

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: panjang_1234
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukkan panjang dan lebar persegi panjang.
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_2034 = input("Masukkan panjang: 5")
lebar_2034 = input("Masukkan lebar: 10")
luas_2034 = int(panjang_2034) * int(lebar_2034)
print("Luas = ", luas_2034)
```

**Penjelasan :**
1. Bagian 1 sampai 2 berfungsi untuk meminta pengguna memasukkan nilai panjang dan lebar persegi panjang menggunakan fungsi input().
2. Kedua input diubah menjadi bilangan integer, lalu dikalikan untuk mendapatkan luas.
3. Hasil luas yang sudah dihitung ditampilkan pada layar.

### 2.2.3 Program "Konstanta_2611532034"

```python
# Buat file dengan nama Konstanta_NIM.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_2034 = float(input('10'))
luas_2034 = PI * jari_2034 * jari_2034
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2034, luas_2034))
```

**Penjelasan :**
1. Program mengimpor Final dan menetapkan PI dengan nilai 3.14 sebagai konstanta.
2. Nilai PI ditampilkan, kemudian pengguna diminta memasukkan jari-jari lingkaran.
3. Nilai jari-jari digunakan bersama PI untuk menghitung luas lingkaran.
4. Hasil perhitungan ditampilkan dengan dua angka di belakang koma.

### 2.2.4 Program "Numerik_2611532034"

```python
# Buat file dengan nama Numerik_NIM.py
# nama variabel ditambah 4 digit terakhir NIM contoh: a_5500
#create a variable with integer value.
a_2034=100
print("The type of variable having value", a_2034, " is ", type (a_2034))

#create a variable with float value.
b_2034=10.2345
print("The type of variable having value", b_2034, " is ", type (b_2034))

#create a variable with complex value.
c_2034=100+3j
print("The type of variable having value", c_2034, " is ", type (c_2034))
```

**Penjelasan :**
1. Bagian 1 sampai 2 berfungsi untuk membuat variabel a_2034 dengan nilai 100 dan menampilkan nilai serta tipe datanya menggunakan fungsi type().
2. Bagian 3 sampai 4 berfungsi untuk membuat variabel b_2034 dengan nilai 10.2345 dan menampilkan nilai serta tipe datanya.
3. Bagian 5 sampai 6 berfungsi untuk membuat variabel c_2034 dengan nilai kompleks 100+3j dan menampilkan nilai serta tipe datanya.
